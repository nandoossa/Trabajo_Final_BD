```python
# CAPA BRONZE: Ingesta usando la ruta exacta en Catalog
tabla = "workspace.bronze.com_014_2025" 

try:
    # Leemos la tabla directamente desde su ubicación en el catálogo 'workspace'
    df_bronze = spark.read.table(tabla)
    
    # Creamos nuestra tabla Delta de trabajo (esta vivirá en el esquema por defecto)
    df_bronze.write.format("delta").mode("overwrite").saveAsTable("bronze_movimientos")
    
    print("✅ ¡Conexión Exitosa! Los datos ya están listos para procesar.")
    display(df_bronze.limit(5))
except Exception as e:
    print(f"❌ Hubo un problema al leer la tabla: {e}")
```
